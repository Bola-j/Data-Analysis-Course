```python
# import all packages and set plots to be embedded inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb

%matplotlib inline
```

```python
# load in the dataset into a pandas dataframe
df = pd.read_csv('diamonds.csv')
```

# 1. Univariate Exploration
## Create 3 different univariate plots

```python
x = df['clarity'].value_counts(sort = False).index
y = df['clarity'].value_counts(sort = False)

plt.bar(x, y)

plt.xlabel('clarity')
plt.ylabel('count')

plt.show()
```

```python
df['color'].unique()
```

```python
plt.figure(figsize=(8, 8))
color_counts = df['color'].value_counts()
plt.pie(color_counts, labels=color_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribution of Color')
plt.show()
```

```python
df
```

```python
plt.figure(figsize=(8, 6))
sb.histplot(data=df, x='price', bins=50)
plt.title('Distribution of Price')
plt.xlabel('price')
plt.ylabel('Frequency')
plt.show()
```
