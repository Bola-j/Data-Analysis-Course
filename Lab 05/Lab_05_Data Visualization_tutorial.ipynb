# There are two main reasons for creating visuals using data:

1. <b>Exploratory analysis "EDA"</b> is done when you are searching for insights. These visualizations don't need to be perfect. You are using plots to find insights, but they don't need to be aesthetically appealing. You are the consumer of these plots, and you need to be able to find the answer to your questions from these plots.

2. <b>Explanatory analysis</b> is done when you are providing your results for others. These visualizations need to provide you the emphasis necessary to convey your message. They should be accurate, insightful, and visually appealing.

**The five steps of the data analysis process:**

1. <b>Extract</b> - Obtain the data from a spreadsheet, SQL, the web, etc.

2. <b>Clean</b> - Here we could use exploratory visuals.

3. <b>Explore</b> - Here we use exploratory visuals.

4. <b>Analyze</b> - Here we might use either exploratory or explanatory visuals.

5. <b>Share</b> - Here is where explanatory visuals live.

# 1. Univariate visualizations 
1.  <b>Bar chart</b> for qualitative variables, for example, the amount (number) of eggs consumed in a meal (categories: {breakfast, lunch, or dinner}). In general, bar chart maps categories to numbers.
2. <b>Pie charts</b> is a common univariate plot type that is used to depict relative frequencies for levels of a categorical variable. A pie chart is preferably used when the number of categories is less, and you'd like to see the proportion of each category.
3. <b>Histograms</b> for quantitative variables. A histogram splits the (tabular) data into evenly sized intervals and displays the count of rows in each interval with bars. A histogram is similar to a bar chart, except that the "category" here is a range of values.

```python
# Necessary imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
from matplotlib import rcParams
%matplotlib inline
```

```python
# Read the csv file, and check its top 10 rows
df = pd.read_csv('pokemon.csv')
print(df.shape)
df.head(10)
```

# 1.1 Bar chart

## Example 1.1.1 Create a vertical bar chart using Matplotlib, with a uniform single color


```python
# Return the Series having unique values
x = df['generation_id'].value_counts(sort=False).index

# Return the Series having frequency count of each unique value
y = df['generation_id'].value_counts(sort=False)

plt.bar(x, y)

# Labeling the axes
plt.xlabel('generation_id')
plt.ylabel('count')

# Dsiplay the plot
plt.show()
```

## Example 1.1.2 Create a vertical bar chart using Seaborn, with default colors


```python
# A semicolon (;) at the end of the statement will supress printing the plotting information
sb.countplot(data=df, x='generation_id');
```

## Example 1.1.3 Create a vertical bar chart using Seaborn, with a uniform single color

```python
sb.color_palette()
```

```python
# The `color_palette()` returns the the current / default palette as a list of RGB tuples. 
# Each tuple consists of three digits specifying the red, green, and blue channel values to specify a color. 
# Choose the first tuple of RGB colors
base_color = sb.color_palette()[0]

# Dynamic-ordering the bars
# The order of the display of the bars can be computed with the following logic.
# Count the frequency of each unique value in the 'generation_id' column, and sort it in descending order
# Returns a Series
freq = df['generation_id'].value_counts()

# Get the indexes of the Series
gen_order = freq.index

# Plot the bar chart in the decreasing order of the frequency of the `generation_id`
sb.countplot(data=df, x='generation_id', color=base_color, order=gen_order);

# Static-ordering the bars
### sb.countplot(data=df, x='generation_id', color=base_color, order=[5,1,3,4,2,7,6]);
```

## Example 1.1.4 Rotate the category labels (not axes)

```python
# Plot the Pokemon type on a Vertical bar chart
sb.countplot(data=df, x='type_1', color=base_color);

# Use xticks to rotate the category labels (not axes) counter-clockwise
plt.xticks(rotation=90);
```

# Bivariate visualizations
1. <b>Scatterplots</b> for quantitative variables. If we want to inspect the relationship between two numeric variables, the standard choice of plot is the scatterplot.
2. <b>Heat Maps</b> is a 2-d version of the histogram that can be used as an alternative to a scatterplot. The variables plotted on each axis can be of any type, whether they take on categorical labels or numeric values. In the latter case, the numeric value must be binned like in a histogram in order to form the grid cells where colors associated with the main variable of interest will be plotted.
3. <b>Box plot</b> is  way of showing the relationship between a numeric variable and a categorical variable.The box plot leans more on the summarization of the data, primarily just reporting a set of descriptive statistics for the numeric values on each categorical level.
4. <b>Clustered Bar Charts</b>

# 2.2 Heat Maps
**Heat maps are useful in the following cases:**

1. To represent a plot for discrete vs. another discrete variable
2. As an alternative to transparency when the data points are enormous

## Example 2.1.1 Default heat plot using Matplotlib.pyplot.hist2d() function

```python
plt.figure(figsize = [18, 6])

# PLOT ON LEFT
plt.subplot(1, 2, 1)
sb.regplot(data = df2, x = 'displ', y = 'comb', x_jitter=0.04, scatter_kws={'alpha':1/10}, fit_reg=False)
plt.xlabel('Displacement (1)')
plt.ylabel('Combined Fuel Eff. (mpg)');

# PLOT ON RIGHT
plt.subplot(1, 2, 2)

# Specify bin edges 
bins_x = np.arange(0.6, 7+0.3, 0.3)
bins_y = np.arange(12, 58+3, 3)

plt.hist2d(data = df2, x = 'displ', y = 'comb', cmin=0.5, cmap='viridis_r', bins = [bins_x, bins_y])
# cmin: The "cmin" parameter specifies the minimum value in a cell before it will be plotted.
# cmap: A list of valid strings can be found here => https://matplotlib.org/stable/api/pyplot_summary.html

plt.colorbar()
plt.xlabel('Displacement (1)')
plt.ylabel('Combined Fuel Eff. (mpg)');
```

# Faceting
One general visualization technique that will be useful for you to know about to handle plots of two or more variables is <b>faceting</b>. In faceting, the data is divided into disjoint subsets, most often by different levels of a categorical variable. For each of these subsets of the data, the same plot type is rendered on other variables. Faceting is a way of comparing distributions or relationships across levels of additional variables, especially when there are three or more variables of interest overall. While faceting is most useful in multivariate visualization, it is still valuable to introduce the technique here in our discussion of bivariate plots.

<b>For example</b>, rather than depicting the relationship between one numeric variable and one categorical variable using a violin plot or box plot, we could use faceting to look at a histogram of the numeric variable for subsets of the data divided by categorical variable levels.

```python
bin_edges = np.arange(12, 58+2, 2)

# Setting col_wrap = 3 means that the plots will be organized into rows of 3 facets each
# Setting sharey = True means that all plots have the same y scale
g = sb.FacetGrid(data = df2, col = 'VClass', col_wrap=3, sharey=True)
g.map(plt.hist, 'comb', bins = bin_edges);
```
