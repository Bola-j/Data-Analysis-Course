# To extract data using web scraping with python, you need to follow these basic steps:
1. Find the URL that you want to scrape.
2. Inspecting the Page.
3. Find the data you want to extract.
4. Write the code.
5. Run the code and extract the data.
6. Store the data in the required format.

```python
import requests
from bs4 import BeautifulSoup
import csv
from itertools import zip_longest
```

```python
url = requests.get('https://wuzzuf.net/search/jobs/?a=spbg&q=python')
```

```python
url
```

```python
html = url.content
```

```python
html
```

```python
try:
	page = BeautifulSoup(html, 'lxml')
except Exception as e:
	print("lxml parser not available, falling back to 'html.parser'. To install lxml run: %pip install lxml")
	page = BeautifulSoup(html, 'html.parser')
print(page)
```

```python
print(page.prettify())
```

```python
all_links = page.find_all('a', {'class':'css-o171kl'})
print(all_links)
```

```python
job_titles = page.find_all('h2', {'class':'css-m604qf'})
company_names = page.find_all('a', {'class':'css-17s97q8'})
company_loc = page.find_all('span', {'class':'css-5wys0k'})

print(job_titles)
print('\n')
print('-' * 114)
print('\n')
print(company_names)
print('\n')
print('-' * 114)
print('\n')
print(company_loc)
```

```python
print(len(job_titles))
print(len(company_names))
print(len(company_loc))
```

```python
for i in range(len(job_titles)):
    print(job_titles[i].text)
    print(company_names[i].text)
    print(company_loc[i].text)
    print('\n')
    print('-' * 50)
    print('\n')
```

```python
jobs = []
c_name = []
c_loc = []
links = []

for i in range(len(job_titles)):
    jobs.append(job_titles[i].text)
    c_name.append(company_names[i].text[:-2])
    c_loc.append(company_loc[i].text.strip())
    links.append(job_titles[i].find('a').attrs['href'])
```

```python
links
```

```python
print(jobs)
print('\n')
print('-' * 50)
print('\n')
print(c_name)
print('\n')
print('-' * 50)
print('\n')
print(c_loc)
print('\n')
print('-' * 50)
print('\n')
print(links)
```

```python
x = [1, 2, 3]
y = ['a', 'b', 'c']
z = ['$', '%', '&']
a = [x, y, z]
print(a)
```

```python
print(*a, sep='')
```

```python
print(list(zip_longest(*a)))
```

```python
data_list = [jobs, c_name, c_loc, links]
print(data_list)
print('\n')
print(*data_list, sep = '----')
print('\n')
print(list(zip_longest(*data_list)))
exported = zip_longest(*data_list)
```

```python
import pandas as pd

df = pd.DataFrame(exported, columns = ['Job Title', 'Company Name', 'Company Location', 'Link'])
```
